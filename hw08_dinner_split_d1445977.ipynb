### 聚餐費用試算器：步驟拆解與說明

在這個試算器中，我們將計算邏輯拆解為以下幾個簡單的步驟：

1. **文字轉數字 (Type Conversion)**：
   我們從輸入或設定得到的餐費與折價券通常是字串（例如 `"1200"`），必須先使用 `int()`（整數）或 `float()`（浮點數）將它們轉換為數值，電腦才能進行數學運算。

2. **計算順序 (Order of Operations)**：
   * **服務費**：依據您的要求，服務費必須以「原始餐費」計算。因此，我們先計算 `原始餐費 * 服務費率`。
   * **總費用**：將`原始餐費` 加上 `服務費`，最後再減去 `折價券金額`。
   * **每人分攤**：將`總費用` 除以 `人數`。

3. **f-string 格式化輸出**：
   我們使用 `f"字串 {變數}"` 的方式來組合文字與變數。如果想要控制小數點位數，可以使用 `{變數:.2f}`，這代表將數值格式化為**小數點後兩位 (float)** 的浮點數。

```python
# 步驟 1：設定原始資料（模擬輸入的字串資料）
original_meal_cost_str = "2500"  # 原始餐費 (字串)
discount_coupon_str = "200"      # 折價券金額 (字串)
service_charge_rate_str = "0.1"  # 服務費率 (字串，0.1 代表 10%)
people_count_str = "4"           # 聚餐人數 (字串)

# 步驟 2：文字轉換為數字
# 使用 int() 將整數型態的字串轉為整數
meal_cost = int(original_meal_cost_str)
discount = int(discount_coupon_str)
people = int(people_count_str)

# 使用 float() 將帶有小數點的字串轉為浮點數
service_rate = float(service_charge_rate_str)

# 步驟 3：計算服務費（以原始餐費計算）
service_fee = meal_cost * service_rate

# 步驟 4：計算總費用（原始餐費 + 服務費 - 折價券）
total_cost = meal_cost + service_fee - discount

# 步驟 5：計算每人分攤金額
cost_per_person = total_cost / people

# 步驟 6：確認總金額是否大於 0（布林值應用）
is_valid_total = total_cost > 0

# 步驟 7：輸出結果與格式化解釋
print("=== 聚餐費用試算結果 ===")
print(f"原始餐費：${meal_cost} 元")
print(f"服務費 (原始餐費的 {service_rate * 100:.0f}%)：${service_fee:.1f} 元")
print(f"折價券折抵：-${discount} 元")
print(f"------------------------")
print(f"總共計算金額：${total_cost:.1f} 元 (計算結果有效：{is_valid_total})")
print(f"每人應付 (共 {people} 人)：${cost_per_person:.2f} 元")

print("\n--- f-string 格式說明 ---")
print("1. {service_rate * 100:.0f} : 將小數乘上 100 後，以小數點後 0 位 (整數) 顯示。")
print("2. {total_cost:.1f} : 代表將總金額顯示到小數點後 1 位。")
print("3. {cost_per_person:.2f} : 代表每人分攤費用顯示到小數點後 2 位，方便計算元與角。")
```

### 聚餐費用試算器 2.0 升級版

在此版本中，我們將解決真實世界中常見的資料清理問題：
1. **字串清理**：去除名稱的頭尾空白、統一轉換為小寫。
2. **金額格式清理**：將包含 `NT$`、逗號與空白的餐費字串（例如 `" NT$1,200 "`）整理為純數字。
3. **預算超支判斷**：使用比較運算子判斷每人分攤費用是否超出預算，並產生布林值結果。
4. **格式化輸出**：使用 f-string 控制小數點位數。

```python
# 原始輸入資料
event_name = "  Weekend Dinner  "  # 聚餐名稱 (含有頭尾空白與大小寫)
meal_amount = " NT$2,500 "         # 原始餐費字串 (含有 NT$、逗號與空白)
discount_coupon_str = "200"        # 折價券金額
service_charge_rate_str = "0.1"    # 服務費率
people_count_str = "4"             # 分攤人數
budget_per_person = 600            # 每人預算 (數值)

# --- 1. 整理聚餐名稱 ---
# strip() 去除頭尾空白，lower() 轉換為小寫
clean_event_name = event_name.strip().lower()

# --- 2. 整理餐費並轉換型別 ---
# 移除 "NT$"、"," 還有空白
clean_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_cost = int(clean_meal_str)  # 轉成整數以利計算

# 轉換其他變數
discount = int(discount_coupon_str)
people = int(people_count_str)
service_rate = float(service_charge_rate_str)

# --- 3. 分步計算費用 ---
# 服務費 (原始餐費 * 服務費率)
service_fee = meal_cost * service_rate

# 應付總額 (原始餐費 + 服務費 - 折價券)
total_cost = meal_cost + service_fee - discount

# 每人費用 (應付總額 / 分攤人數)
cost_per_person = total_cost / people

# --- 4. 判斷是否超過預算 ---
# 若每人費用大於預算，則結果為 True；否則為 False
is_over_budget = cost_per_person > budget_per_person

# --- 5. 輸出費用摘要 ---
print("========================================")
print(f"聚餐名稱: {clean_event_name}")
print("========================================")
print(f"原始餐費: ${meal_cost:.2f} 元")
print(f"服務費 ({service_rate * 100:.0f}%): ${service_fee:.2f} 元")
print(f"折價券折抵: -${discount:.2f} 元")
print(f"----------------------------------------")
print(f"應付總額: ${total_cost:.2f} 元")
print(f"分攤人數: {people} 人")
print(f"每人費用: ${cost_per_person:.2f} 元")
print(f"每人預算: ${budget_per_person:.2f} 元")
print(f"是否超過預算: {is_over_budget}")
print("========================================")
```
